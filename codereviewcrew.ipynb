```python
from crewai import Agent,LLM
from google.colab import userdata

gemini_api_key = userdata.get('sec1')


llm = LLM(
   model="gemini/gemini-3.5-flash",
   api_key=gemini_api_key
)

Code_Analyzer = Agent(
    role="Code_Analyzer",
    goal="Analyze the given buggy code, understand its intended functionality, and identify all potential issues",
    backstory=
      "You are an experienced software architect."
      "You excel at reading and understanding code to identify issues including:"
      "- syntax errors, logical errors, and runtime errors"
      "- incorrect variable usage and scope issues"
      "- missing edge cases and boundary conditions"
      "Keep your analysis clear and structured so the debugger can fix this issue.",
    verbose=True,
    llm=llm,
)
```

```python
Senior_Debugger = Agent(
    role="Senior_Debugger",
    goal="Fix all identified bugs and produce clean, working code based on the analysis report",
    backstory=
      "You are a senior software engineer specialized in debugging and fixing code."
      "You write clean, corrected code with:"
      "- Proper fix for each identified bug"
      "- Comments explaining what was wrong and how it was fixed"
      "- Improved variable names and code structure where needed"
      "You always produce a complete, runnable fixed versio",
    verbose=True,
    llm=llm
)
```

```python
code_reviewer = Agent(
    role="code_reviewer",
    goal="Verify the fixed code is correct, test edge cases, and ensure it matches the intended functionality",
    backstory=
      "You are a meticulous code reviewer and tester."
      "You carefully check:"
      "- Does the fixed code run without errors?"
      "- Are ALL the identified bugs actually fixed?"
      "- Does it handle edge cases properly?"
      "- Code quality, readability, and best practices"
      "Suggest final improvements and output the FINAL",
    verbose=True,
    llm=llm
)
```

```python
from crewai import Task
```

```python
task_analyze = Task(
    description=
  "Take the user's buggy code: {buggy_code}"
  "1. Read and understand what the code is supposed to do"
  "2. Identify ALL bugs: syntax errors, logical errors, runtime errors"
  "3. List each bug with its line number and explanation"
  "Output format:"
  "## Bug Analysis Report"
  "- Intended Functionality: ..."
  "- Language: ..."
  "- Bug 1: [Line X] - Description of the issue"
  "- Bug 2: [Line X] - Description of the issue"
  "- Severity: ..."
  "- Root Cause Summary: ...",
    expected_output="A clear markdown Bug Analysis Report listing all identified issues",
    agent=Code_Analyzer
)
```

```python
task_fix = Task(
    description=
  "Using the bug analysis report from the previous task"
  "Fix ALL the identified bugs and produce clean, working code."
  "- Fix each bug mentioned in the analysis report"
  "- Add a comment above each fix explaining: # FIX: what was wrong -> what was changed"
  "- Ensure the code runs without any errors"
  "- Maintain the original code structure as much as possible"
  "- Final answer MUST be ONLY the complete fixed Python code and a summary of all fixes made",
    expected_output="A complete, runnable fixed Python script with fix comments",
    agent=Senior_Debugger
)
```

```python
task_review =Task(
    description=
  "Review the fixed code from the previous task."
  "1. Verify each bug from the analysis report is actually fixed"
  "2. Test mentally with normal inputs and edge cases"
  "3. Check for any new bugs introduced during fixing"
  "4. Verify code quality and readability"
  "5. Output the FINAL, verified, bug-free code"
  "Your final answer MUST be ONLY the complete verified Python code along with:"
  "- A checklist of all bugs fixed"
  "- Edge cases tested"
  "- Any additional improvements made",
    expected_output="Final verified, bug-free Python script with fix checklist and test summary",
    agent=code_reviewer
)
```

```python
from crewai import Crew, Process

code_debug_crew = Crew(
      agents=[Code_Analyzer,Senior_Debugger,code_reviewer],
      tasks=[task_analyze, task_fix, task_review],
      process=Process.sequential,
      verbose=True
)
```

```python

import asyncio

game_idea = """
def calculate_average(numbers):
    total = 0
    for i in range(len(numbers)):
        total += numbers[i + 1]
    avg = total / len(numbers)
    return avg

def find_max(numbers):
    max_val = 0
    for num in numbers:
        if num > max_val:
            max_val = num
    return max_val

def remove_duplicates(lst):
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[i] == lst[j]:
                lst.remove(lst[j])
    return lst

numbers = [3, 7, 2, 8, 2, 7, 4]
print("Average:", calculate_average(numbers))
print("Maximum:", find_max(numbers))
print("Without duplicates:", remove_duplicates(numbers))
"""

```

```python
result = await code_debug_crew.kickoff_async(
    inputs={"buggy_code": game_idea}
)

print(result)
```
